### Chains In LangChain

```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage 
from langchain_openai import ChatOpenAI
import os

my_message = [
    SystemMessage(content= "You are a helpful assistant that provides fun facts."),
    HumanMessage(content= "Hello, tell me a fun fact")
]

llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.9, api_key= os.environ.get("API_KEY"))

llm.invoke(my_message).content
```

First Chain

```python
#Task 1 - Prompt

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import HumanMessage, SystemMessage

prompt_message = ChatPromptTemplate.from_messages([
    SystemMessage(content= "You are a helpful assistant that provides fun facts."),
    HumanMessage(content= "{input}")
])


```

```python
### TASK 2 -  llm

llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.9, api_key= os.environ.get("API_KEY"))
```

```python
### TASK 3 -  String Output Parser

from langchain_core.output_parsers import StrOutputParser

str_parser = StrOutputParser()

```

## Chain Invocation

```python
chain = prompt_message | llm | str_parser

chain.invoke({"input" : "Hello, tell me a fun fact about space."})
```

```python
from langchain_core.runnables import RunnableSequence

chain_run = RunnableSequence(prompt_message, llm, str_parser)

chain_run.invoke({"input" : "Hello, tell me a fun fact about space."})
```
