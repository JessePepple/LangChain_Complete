```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage 
from langchain_openai import ChatOpenAI
import os

my_message = [
    SystemMessage(content= "You are a helpful assistant that provides fun facts."),
    HumanMessage(content= "Hello, tell me a fun fact")
]

llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.9, api_key= os.environ.get("API_KEY"))

llm.invoke(my_message).content
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

chat_prompt = ChatPromptTemplate.from_messages(
[
    SystemMessage(content= "You are a helpful assistant that provides fun facts."),
    HumanMessage(content= "{input}")
]
)
```

```python
llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.9, api_key= os.environ.get("API_KEY"))
```

```python
from langchain_core.output_parsers import StrOutputParser

str_parser = StrOutputParser()
```

```python
### Custom Functions
from langchain_core.runnables import RunnableLambda
def dictionary_maker(text:str)-> dict:
    return {"output" : text}


dictionary_maker_runnable = RunnableLambda(dictionary_maker)
```

```python
### Template for post

prompt_post = ChatPromptTemplate.from_messages(
[
    SystemMessage(content= "You are a helpful assistant that provides fun facts."),
    HumanMessage(content= "{input}")
])
```

```python
chain = chat_prompt | llm | str_parser | dictionary_maker_runnable

chain.invoke({"input" : "Hello, who is Naruto."})
```

### Parallel Chains

```python
prompt_template = ChatPromptTemplate.from_messages(
[
    SystemMessage(content= "You are a helpful assistant that provides fun facts."),
    HumanMessage(content= "Create a post for LinkedIn: {input}")
]
)

llm_openai = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.9, api_key= os.environ.get("API_KEY"))

str_parser = StrOutputParser()

chain_linked = prompt_template | llm_openai | str_parser 

chain_linked.invoke({"input" : "I just learned about LangChain, a framework for building applications with LLMs. It's really cool!"})
```
