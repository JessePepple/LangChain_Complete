```python
import os
from dotenv import load_dotenv

if os.environ.get("API_KEY") is None:
    raise ValueError("API_KEY environment variable is not set. Please set it in the .env file.")

else:
    print("API_KEY is set. Proceeding with the application.")

```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model= "gpt-5-nano", temperature=0, api_key=os.environ.get("API_KEY"))

response = llm.invoke("What is the capital of France? and what is the population of France?")
print(response.content)
```

```python
from langchain.chat_models import init_chat_model

llm_openai = init_chat_model(model= "gpt-5-nano", temperature=0, api_key=os.environ.get("API_KEY"))

response_model = llm_openai.invoke("What is the capital of France? and what is the population of France?")
print(response_model.content)
```
