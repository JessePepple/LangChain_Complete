```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage 
from langchain_openai import ChatOpenAI
import os

my_message = [
    SystemMessage(content= "You are a helpful assistant that provides fun facts."),
    HumanMessage(content= "Hello, tell me a fun fact")
]

llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.9, api_key= os.environ.get("API_KEY"))

llm.invoke(my_message).content
```

## Pydantic 

```python
from pydantic import BaseModel, Field

class llm_schema(BaseModel):
    setup: str = Field(..., description="The setup of the joke")
    punchline: str = Field(..., description="The punchline of the joke")

```

```python
llm_strcutured_output = llm.with_structured_output(llm_schema)


llm_strcutured_output.invoke("Tell me a joke about programming")
```

Using Typed Dict

```python
from typing import TypedDict

class llm_joke(TypedDict):
    setup: str
    punchline: str
```

```python
obj = llm_joke({"setup" : "some setup", "punchline" : "some punchline"})

obj
```
